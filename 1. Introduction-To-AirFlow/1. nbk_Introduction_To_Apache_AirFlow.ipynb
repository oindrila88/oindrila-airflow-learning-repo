# What is a <b>Workflow</b>?
* A <b>Workflow</b> is a <b>sequence</b> of <b>tasks</b>, which starts on a <b>schedule</b> or gets <b>triggered</b> by an event.
* <b>Workflow</b>s are frequently used to handle <b>big data processing pipelines</b>.
* Following is an example of a typical <b>Workflow</b> -
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-airflow-learning-repo/images/typical_workflow.jpg'>
  * <b>Task 1</b>: The first task of the <b>Workflow</b> is to download data from a source.
  * <b>Task 2</b>: The second task of the <b>Workflow</b> is to send the downloaded data somewhere else to process.
  * <b>Task 3</b>: The third task of the <b>Workflow</b> is to monitor the data processing step to check if it is completed.
  * <b>Task 4</b>: Once the data processing step is completed, the fourth task of the <b>Workflow</b> is to fetch the result of the data processing and generate a report.
  * <b>Task 5</b>: The fifth task of the <b>Workflow</b> is to send the generated report by email.

# A Traditional <b>ETL</b> Approach
* In a traditional <b>ETL</b> approach, suppose, the requirement is to get the business data out of a <b>Relational Database</b>, like <b>Oracle</b> and load the data into a <b>Data Lake</b> storage, like <b>S3</b>, which will be then processed by <b>Databricks</b>.
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-airflow-learning-repo/images/traditional_etl_approach.jpg'>
* A naive approach to achieve this would be to write a <b>script</b>, in which multiple <b>task</b>s can be specified.
* The <b>task</b>s can be the following -
  * The first <b>task</b> can be to connect to the <b>Oracle</b> database.
  * The second <b>task</b> can be to pull data from the <b>Oracle</b> database.
  * The third <b>task</b> can be to send the data to <b>S3</b> bucket.
  * The fourth <b>task</b> can be to connect to <b>Databricks</b>.
  * The fifth <b>task</b> can be to call a <b>Databricks Workflow</b> to process the data in the <b>S3</b> bucket.
* It is also possible to schedule the <b>script</b> as a <b>cron job</b> based on the need, like - <b>daily</b>, <b>hourly</b> etc.

# Problems of the Traditional <b>ETL</b> Approach
* There are multiple problems with the traditional <b>ETL</b> approach -
  * <b>1</b>. <b>Handling Failures</b>: It is better to retry a <b>task</b> for which the entire <b>ETL</b> pipeline fails. But, retrying a <b>task</b> will need the below information -
    * For how many times a <b>task</b> can retry if it fails.
    * At what interval a <b>task</b> can retry if it fails.
  * <b>2</b>. <b>Monitoring</b>: There should be a mechanism to keep track of the <b>status</b> of each <b>task</b>s in an <b>ETL</b> pipeline.
  <br>The mechanism should also check how long each of the <b>task</b>s would run to check if any <b>task</b> runs for too long.
  * <b>3</b>. <b>Dependencies</b>:
    * <b> Data Dependency</b>: There should be a mechanism to handle the data dependency issue, i.e., in case of an <b>ETL</b> pipeline, if the <b>upstream data</b> goes missing, then, it would be desired to not run the <b>downstream task</b>s.
    * <b>Execution Dependency</b>: There should be a mechanism to handle the execution dependency issue, i.e., handling an <b>ETL</b> pipeline that consists of multiple <b>cron jobs</b>, and, where one of the <b>cron jobs</b> takes too long to run.
    <br>Example - There are two <b>cron jobs</b> in an <b>ETL</b> pipeline. <b>Job1</b> takes an hour to finish. Suppose, <b>Job1</b> is scheduled at <b>4PM IST</b>. So, it is expected to finish by <b>5PM IST</b>.
    <br>Hence, <b>Job2</b> is scheduled at <b>5:30PM IST</b> to have a buffer of <b>30 minutes</b>.
    <br>But, suppose, at one day <b>Job1</b> takes two hours to finish. So, that day, <b>Job2</b> started before <b>Job1</b> was finished. This is the execution dependency issue.
  * <b>4</b>. <b>Scalability</b>: Suppose, a compamy uses one machine to schedule one <b>cron job</b>. So, if more <b>cron jobs</b> are to be scheduled, that company will scale up more machines to support the number of <b>cron jobs</b>. But, after a time, it will not be possible to scale up more machines.
  <br>Also, in such cases, if one <b>cron job</b> would be dependent on another <b>cron job</b>, then there would be no central scheduler between the multiple machines to schedule one <b>cron job</b> after the other.
  * <b>5</b>. <b>Deployment</b>: There should also be a mechanism to keep track of, or, maintain all the new changes that constantly happen.
  * <b>6</b>. <b>Process Historic Data</b>: It is the regular need of all the companies that uses big data to backfill, or, rerun the historical data to generate a report.
  <br>Such report can be used to compare the data of the last six months for that company to see any upward or downward trend.

# Why to Use <b>Apache AirFlow</b>?
* To handle all the problems of the traditional <b>ETL</b> approach nicely, the <b>Apache AirFlow</b> is used.
* <b>Apache AirFlow</b> is a platform to programmatically author, schedule, and, monitor <b>workflow</b>s, or, <b>data pipeline</b>s, and, is developed by <b>Airbnb</b>.
  * <b>Apache AirFlow</b> takes care of the <b>execution dependency</b> issue by letting the users write all the <b>task</b>s, and, the inter-dependencies among the <b>task</b>s in an <b>ETL</b> pipeline in <b>Python</b>.
  * <b>Apache AirFlow</b> takes care of the <b>scalability</b> issue by executing, scheduling, and, distributing the <b>task</b>s across the <b>Worker Nodes</b> of the <b>Cluster</b>.
  * <b>Apache AirFlow</b> takes care of the <b>monitoring</b> issue by displaying a view of the present and the past runs of an <b>ETL</b> pipeline, and, providing a <b>logging</b> feature.
  * <b>Apache AirFlow</b> is extensible through <b>plugins</b>.
  * <b>Apache AirFlow</b> interacts well with all the major <b>Cloud</b>s, or, <b>Database</b> systems.
